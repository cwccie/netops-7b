This notebook walks through the synthetic data generation pipeline for NetOps-7B training.

## Overview

The pipeline generates 15K-30K instruction/response pairs across three categories:
1. **Configuration Generation** — Cisco, Juniper, Arista device configs
2. **Troubleshooting** — Network issue diagnosis and resolution
3. **Command Explanation** — CLI command reference with examples

```python
from netops_7b.data.curate import DatasetCurator
from netops_7b.data.formats import AlpacaFormatter, ShareGPTFormatter
from netops_7b.data.quality import QualityFilter

# Generate dataset
curator = DatasetCurator(target_size=20000, seed=42)
pairs = curator.generate()
print(f"Generated {len(pairs)} examples")

# Show stats
stats = curator.stats(pairs)
for key, value in stats.items():
    print(f"  {key}: {value}")
```

```python
# Quality filtering
qf = QualityFilter()
filtered = qf.filter(pairs)
print(f"After filtering: {len(filtered)} examples")
print(f"Rejection stats: {qf.stats()}")
```

```python
# Convert to ShareGPT format and save
formatter = ShareGPTFormatter()
formatter.save(filtered, "data/train_sharegpt.json")
print("Saved to data/train_sharegpt.json")
```

```python
# Preview examples
for pair in filtered[:3]:
    print(f"=== [{pair.category}] [{pair.vendor}] ===")
    print(f"Instruction: {pair.instruction[:100]}...")
    print(f"Response: {pair.response[:100]}...")
    print()
```
